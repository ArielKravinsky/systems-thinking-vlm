# Supervisor Report: vec_both Optimization Findings

This notebook summarizes the exploratory optimization that identified the best Spearman correlation under rm<=3 image removals.


## Key Finding
Best setting: remove images **3,4,5** with **min_answers=4**.
- n=41, Spearman r=0.311898, p=0.0471229, FDR=0.602686.
This is an exploratory finding; FDR is not significant.


```python
import pandas as pd
from pathlib import Path

root = Path.cwd()
res = root / 'results'
stamp = '20260417'
sweep = pd.read_csv(res / f'maxcorr_q1ge65_vecboth_spearman_all_rm3_{stamp}.csv')
sweep = sweep.dropna(subset=['spearman_r', 'spearman_p'])
sweep = sweep[sweep['n_subjects'] >= 3].copy()
best = sweep.sort_values(['spearman_r','spearman_p'], ascending=[False, True]).iloc[0]
best
```

```python
baseline = sweep[sweep['removed_images'].fillna('') == '']
best_no_rm = baseline.sort_values(['spearman_r','spearman_p'], ascending=[False, True]).iloc[0]
pd.DataFrame([best_no_rm, best])[['removed_images','min_answers','n_subjects','spearman_r','spearman_p','fdr_bh']]
```

```python
sel = sweep[sweep['removed_images'] == '3;4;5'].sort_values('min_answers')
sel[['min_answers','n_subjects','spearman_r','spearman_p','fdr_bh']]
```

```python
inc_main = pd.read_csv(res / f'included_subjects_q1ge65_with_image_vecboth_{stamp}.csv')
inc_best = pd.read_csv(res / f'included_subjects_best_rm3_345_min4_vecboth_{stamp}.csv')
print('Main cohort size:', len(inc_main))
print('Best config size:', len(inc_best))
inc_best[['subject_id','questionnaire_1_score','image_questionnaire_score_vec_both']].head(15)
```
